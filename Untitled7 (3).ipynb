Renaming column's names, to make them brief and clear.
Also removing incomplete responses(in last 2 columns there are answers like - and . i changed them to no).

```python
rename_map = {
    'Отметка времени': 'timestamp',
    'What is your age group?': 'age_group',
    'What is your role?': 'role',
    'At what age did you start using a smartphone?  ': 'smartphone_age',
    'How would you describe your social media usage?  ': 'social_media_usage',
    'I feel that the content recommended to me influences my opinions or choices.  ': 'influence_recommend',
    'I often check social media without a specific reason.  ': 'check_no_reason',
    'I feel uneasy when I cannot access social media for a while.  ': 'uneasy_no_access',
    'I notice that my beliefs on social issues are influenced by posts I see frequently.  ': 'beliefs_influenced',
    'Sometimes I accept information on social media without checking other sources.  ': 'accept_info_no_check',
    'I pay more attention to content that evokes strong emotions, even if it’s not informative.  ': 'attention_emotion',
    'I notice differences in the content, tone, or style of posts written in different languages (Kazakh, Russian, English) on social media.  ': 'lang_diff_notice',
    'I often check social media immediately after receiving a notification, even if I’m busy.  ': 'check_notification',
    'I find it difficult to focus on long posts or articles after spending time on social media.  ': 'focus_long_posts',
    'I feel that the language of available content online affects what I watch or read.  ': 'lang_affects_content',
    'Which types of content do you engage with most on social media?  ': 'content_types',
    'When you see a post you strongly disagree with, what do you usually do?  ': 'disagree_action',
    'Have you noticed any changes in the language, topics, or tone of content you consume over the past few years? If so, what changes?  ': 'changes_language_topics_tone',
    'Do you feel that social media has influenced your attention span, memory, or way of processing information?  ': 'influence_attention_memory'
}

df.rename(columns=rename_map, inplace=True)

def clean_last_columns(value):
    if pd.isna(value):
        return 'no'
    value_str = str(value).strip()
    if value_str in ['', '-', '.']:
        return 'no'
    return value_str

df['changes_language_topics_tone'] = df['changes_language_topics_tone'].apply(clean_last_columns)
df['influence_attention_memory'] = df['influence_attention_memory'].apply(clean_last_columns)

required_cols = ['age_group', 'role', 'social_media_usage', 'influence_recommend']
df_clean = df.dropna(subset=required_cols, how='any')

df_clean.to_csv('survey2_cleaned.csv', index=False)

df.head()
```

Hypothesis Testing (at least two tests)

```python
# Hypothesis Test 1: Correlation between social media usage intensity and attention problems
print("\n" + "="*60)
print("TEST 1: Correlation Analysis")
print("="*60)
print("Hypothesis: Higher social media usage is positively correlated with")
print("difficulty focusing on long posts (focus_long_posts)")

# Encode social media usage as numeric
usage_map = {
    'Very active user': 4,
    'Moderately active user': 3,
    'Occasional user': 2,
    'Rarely use social media': 1,
    'I do not use social media': 0
}
df['usage_numeric'] = df['social_media_usage'].map(usage_map)

# Remove NaN values
corr_data = df[['usage_numeric', 'focus_long_posts']].dropna()
corr, p_value = pearsonr(corr_data['usage_numeric'], corr_data['focus_long_posts'])
print(f"\nCorrelation between social media usage intensity and focus difficulty: {corr:.3f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print(f"\n✓ Result: Significant positive correlation (p < 0.05)")
    print(f"  Interpretation: As social media usage increases, difficulty focusing on long content increases")
else:
    print(f"\n✗ Result: No significant correlation (p >= 0.05)")
```

```python
# Hypothesis Test 2: t-test comparing Very active vs Occasional users on attention span
print("\n" + "="*60)
print("TEST 2: Independent t-test")
print("="*60)
print("Hypothesis: Very active users report significantly higher difficulty focusing")
print("on long posts compared to occasional users")

# Separate groups
very_active = df[df['social_media_usage'] == 'Very active user']['focus_long_posts'].dropna()
occasional = df[df['social_media_usage'] == 'Occasional user']['focus_long_posts'].dropna()

print(f"\nVery active users (n={len(very_active)}): mean = {very_active.mean():.2f}, sd = {very_active.std():.2f}")
print(f"Occasional users (n={len(occasional)}): mean = {occasional.mean():.2f}, sd = {occasional.std():.2f}")

# Perform t-test
t_stat, p_value = ttest_ind(very_active, occasional)
print(f"\nT-statistic: {t_stat:.3f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print(f"\n✓ Result: Significant difference (p < 0.05)")
    print(f"  Interpretation: Very active users report significantly more difficulty")
    print(f"  focusing on long content than occasional users")
else:
    print(f"\n✗ Result: No significant difference (p >= 0.05)")
```
