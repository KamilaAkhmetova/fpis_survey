```python
df = pd.read_csv('survey2.csv')
```

```python
rename_map = {
    'Отметка времени': 'timestamp',
    'What is your age group?': 'age_group',
    'What is your role?': 'role',
    'At what age did you start using a smartphone?  ': 'smartphone_age',
    'How would you describe your social media usage?  ': 'social_media_usage',
    'I feel that the content recommended to me influences my opinions or choices.  ': 'influence_recommend',
    'I often check social media without a specific reason.  ': 'check_no_reason',
    'I feel uneasy when I cannot access social media for a while.  ': 'uneasy_no_access',
    'I notice that my beliefs on social issues are influenced by posts I see frequently.  ': 'beliefs_influenced',
    'Sometimes I accept information on social media without checking other sources.  ': 'accept_info_no_check',
    'I pay more attention to content that evokes strong emotions, even if it’s not informative.  ': 'attention_emotion',
    'I notice differences in the content, tone, or style of posts written in different languages (Kazakh, Russian, English) on social media.  ': 'lang_diff_notice',
    'I often check social media immediately after receiving a notification, even if I’m busy.  ': 'check_notification',
    'I find it difficult to focus on long posts or articles after spending time on social media.  ': 'focus_long_posts',
    'I feel that the language of available content online affects what I watch or read.  ': 'lang_affects_content',
    'Which types of content do you engage with most on social media?  ': 'content_types',
    'When you see a post you strongly disagree with, what do you usually do?  ': 'disagree_action',
    'Have you noticed any changes in the language, topics, or tone of content you consume over the past few years? If so, what changes?  ': 'changes_language_topics_tone',
    'Do you feel that social media has influenced your attention span, memory, or way of processing information?  ': 'influence_attention_memory'
}
df.rename(columns=rename_map, inplace=True)

def clean_last_columns(value):
    if pd.isna(value):
        return 'no'
    value_str = str(value).strip()
    if value_str in ['', '-', '.']:
        return 'no'
    return value_str

df['changes_language_topics_tone'] = df['changes_language_topics_tone'].apply(clean_last_columns)
df['influence_attention_memory'] = df['influence_attention_memory'].apply(clean_last_columns)

required_cols = ['age_group', 'role', 'social_media_usage', 'influence_recommend']
df_clean = df.dropna(subset=required_cols, how='any')

df_clean.to_csv('survey2_cleaned.csv', index=False)

df.head()
```

```python
negative_keywords = [
    'war', 'poverty', 'injustice', 'depressed', 'too much', 'politics',
    'political', 'anxiety', 'stress', 'overload', 'hard to concentrate',
    'attention span', 'memory', 'shorter', 'visual', 'quick trends',
    'unnecessary', 'degenerate', 'brainrot', 'polarized', 'black and white',
    'commodity', 'radical', 'slop', 'aggressive', 'uncomfortable', 'anxious',
    'тревог', 'перегрузк', 'внимани', 'концентр', 'политик', 'депрес'
]
```

```python
def classify_sentiment(text):
    """
    Определяет тональность ответа:
    0 = Negative (проблемный ответ - есть жалоба, тревога, перегрузка)
    1 = Positive (нормальный/нейтральный ответ)
    """
    if pd.isna(text) or text == 'no' or text == 'idk' or text == '.' or text == '-':
        return 1  # Positive (нет проблемы)

    text_lower = str(text).lower()

    for keyword in negative_keywords:
        if keyword in text_lower:
            return 0  # Negative (проблемный)

    return 1

df['actual_sentiment'] = df['changes_language_topics_tone'].apply(classify_sentiment)
```

```python
df.head()
```

```python
def simple_model_predict(text):
    """
    Простая модель на основе ключевых слов
    Имитирует работу ML-модели с некоторыми ошибками
    """
    text_lower = str(text).lower()

    problem_indicators = ['politics', 'political', 'shorter', 'visual', 'war', 'depressed', 'тревог']

    score = 0
    for word in problem_indicators:
        if word in text_lower:
            score += 1

    if score >= 2:
        return 0  # Negative
    elif score == 1:
        return 0 if np.random.random() > 0.3 else 1
    else:
        return 1  # Positive
```

```python
np.random.seed(42)
df['predicted_sentiment'] = df['changes_language_topics_tone'].apply(simple_model_predict)
```

```python
print(df[['changes_language_topics_tone', 'actual_sentiment', 'predicted_sentiment']].head(10))
```

```python
actual = df['actual_sentiment'].values
predicted = df['predicted_sentiment'].values

cm = confusion_matrix(actual, predicted)
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['Negative (проблемный)', 'Positive (нормальный)']
)
disp.plot(cmap='Blues')
plt.title('Confusion Matrix - Classification of Social Media Responses')
plt.show()
```

```python
actual = df['actual_sentiment'].values
predicted = df['predicted_sentiment'].values

TP = sum((a == 0 and p == 0) for a, p in zip(actual, predicted))
FP = sum((a == 1 and p == 0) for a, p in zip(actual, predicted))
FN = sum((a == 0 and p == 1) for a, p in zip(actual, predicted))
TN = sum((a == 1 and p == 1) for a, p in zip(actual, predicted))
```

```python
precision = TP / (TP + FP) if (TP + FP) > 0 else 0
recall = TP / (TP + FN) if (TP + FN) > 0 else 0
f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

print("\n" + "=" * 60)
print("CLASSIFICATION METRICS")
print("=" * 60)
print(f"\nPrecision = {TP} / ({TP} + {FP}) = {precision:.4f} ({precision*100:.2f}%)")
print(f"Recall    = {TP} / ({TP} + {FN}) = {recall:.4f} ({recall*100:.2f}%)")
print(f"F1-score  = {f1:.4f} ({f1*100:.2f}%)")
```

```python
# checking
precision_sk = precision_score(actual, predicted, pos_label=0)
recall_sk = recall_score(actual, predicted, pos_label=0)
f1_sk = f1_score(actual, predicted, pos_label=0)

print("\n" + "=" * 60)
print("CLASSIFICATION METRICS")
print("=" * 60)
print(f"\nSklearn Precision: {precision_sk:.4f}")
print(f"Sklearn Recall:    {recall_sk:.4f}")
print(f"Sklearn F1-score:  {f1_sk:.4f}")
```

```python
import pandas as pd
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
from deep_translator import GoogleTranslator
import matplotlib.pyplot as plt

df_survey = pd.read_csv('survey2.csv')

# Переименование (из твоего кода)
rename_map = {
    'Have you noticed any changes in the language, topics, or tone of content you consume over the past few years? If so, what changes?  ': 'changes_language_topics_tone',
}
df_survey.rename(columns=rename_map, inplace=True)

# Убираем пустые
responses = df_survey['changes_language_topics_tone'].dropna()
responses = responses[~responses.isin(['-', '.', 'no', 'idk', ''])]
print(f'Содержательных ответов: {len(responses)}')
print('\nПримеры:')
print(responses.head(10).to_string())
```

```python
# Ключевые слова для разделения "раньше" vs "сейчас"
PAST_KEYWORDS = [
    'before', 'used to', 'previously', 'earlier', 'in the past',
    'at first', 'раньше', 'раньше было', 'прежде', 'до этого',
    'бұрын', 'originally', 'back then', 'few years ago', 'once'
]

PRESENT_KEYWORDS = [
    'now', 'currently', 'today', 'recently', 'nowadays', 'these days',
    'at the moment', 'сейчас', 'теперь', 'в последнее время',
    'more and more', 'increasingly', 'has become', 'have become',
    'started to', 'begun to', 'more often', 'getting more'
]

analyzer = SentimentIntensityAnalyzer()

def extract_temporal_sentiment(text):
    """
    Извлекаем sentiment из частей текста которые говорят про прошлое vs настоящее.
    Возвращает словарь с оценками.
    """
    if not isinstance(text, str) or len(text) < 10:
        return None
    
    text_lower = text.lower()
    sentences = text.replace('!', '.').replace('?', '.').split('.')
    sentences = [s.strip() for s in sentences if len(s.strip()) > 5]
    
    past_sentences = []
    present_sentences = []
    
    for sent in sentences:
        sent_lower = sent.lower()
        if any(kw in sent_lower for kw in PAST_KEYWORDS):
            past_sentences.append(sent)
        elif any(kw in sent_lower for kw in PRESENT_KEYWORDS):
            present_sentences.append(sent)
    
    result = {
        'full_text': text,
        'has_temporal': len(past_sentences) > 0 or len(present_sentences) > 0,
        'past_text': ' '.join(past_sentences) if past_sentences else None,
        'present_text': ' '.join(present_sentences) if present_sentences else None,
        'past_sentiment': None,
        'present_sentiment': None,
        'overall_sentiment': analyzer.polarity_scores(text)['compound']
    }
    
    if past_sentences:
        result['past_sentiment'] = analyzer.polarity_scores(' '.join(past_sentences))['compound']
    if present_sentences:
        result['present_sentiment'] = analyzer.polarity_scores(' '.join(present_sentences))['compound']
    
    return result

# Применяем
results = []
for text in responses:
    r = extract_temporal_sentiment(str(text))
    if r:
        results.append(r)

df_temporal = pd.DataFrame(results)

# Смотрим что получилось
temporal_df = df_temporal[df_temporal['has_temporal'] == True]
print(f'Ответов с временной перспективой: {len(temporal_df)} из {len(df_temporal)}')
print(f'\nСредний sentiment (прошлое): {temporal_df["past_sentiment"].mean():.4f}')
print(f'Средний sentiment (настоящее): {temporal_df["present_sentiment"].mean():.4f}')
print(f'Средний sentiment (весь ответ): {df_temporal["overall_sentiment"].mean():.4f}')

print('\n--- Примеры с временной перспективой ---')
for _, row in temporal_df.head(5).iterrows():
    print(f'\nТекст: {row["full_text"][:100]}...')
    if row['past_text']:
        print(f'  ПРОШЛОЕ ({row["past_sentiment"]:.2f}): {row["past_text"][:80]}')
    if row['present_text']:
        print(f'  НАСТОЯЩЕЕ ({row["present_sentiment"]:.2f}): {row["present_text"][:80]}')
```
